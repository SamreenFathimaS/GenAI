```python
#Synchronous Version
import time
def soru():
    print ("starting the soru cooking process")
    time.sleep(2)
    print("soru is now cooking...")
    time.sleep(5)
    print("soru is ready!")
def kolambu():
    print("starting the kolambu preparation..")
    time.sleep(2)
    print("kolambu is now simmering..")
    time.sleep(5)
    print("kolambu is ready!")
start = time.time()
soru = soru()
kolambu = kolambu()
end = time.time()
print(f"time:{end-start}minutes")
```

```python
#ASynchronous Version
import time
import asyncio
async def soru():
    print ("starting the soru cooking process")
    await asyncio.sleep(2)
    print("soru is now cooking...")
    await asyncio.sleep(5)
    print("soru is ready!")
async def kolambu():
    print("starting the kolambu preparation..")
    await asyncio.sleep(2)
    print("kolambu is now simmering..")
    await asyncio.sleep(5)
    print("kolambu is ready!")
async def main():
   start = time.time()
   await asyncio.gather(
     soru(),
     kolambu()
 )
   end = time.time()
   print(f"time:{end-start}minutes")
await main()
```
