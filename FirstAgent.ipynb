```python
#Tools in AutoGen for calendar
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from dotenv import load_dotenv
import os
import asyncio
from datetime import date
load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
model_client = OpenAIChatCompletionClient(model='gpt-4.1', api_key=api_key)
# Dummy calendar data
calendar_data = {
    "2025-08-16": ["Team meeting at 10 AM", "Doctor appointment at 5 PM"],
    "2025-08-18": ["Project deadline"],
    "2025-08-20": []
}

# Tool function
def get_calendar(input_date: str = None) -> str:
    """
    Return if there are events/meetings on the given date.
    If no input_date is provided, defaults to today.
    """
    if not input_date:
        input_date = str(date.today())

    events = calendar_data.get(input_date, [])

    if not events:
        return f"No events or meetings on {input_date}."
    else:
        return f"Yes, you have {len(events)} event(s) on {input_date}."

# Create assistant agent
assistant = AssistantAgent(
    name="Chand",
    model_client=model_client,
    system_message=(
        "You are a calendar assistant. "
        "Use the get_calendar tool to check if there are any events or meetings. "
        "If the user asks about 'today', use today's date. "
        "If the user gives a specific date, pass that date (YYYY-MM-DD) to the tool."
    ),
    tools=[get_calendar],
)

# Example runs (async)
result1 = await assistant.run(task="Do I have any meeting today?")
result2 = await assistant.run(task="Do I have meetings on 2025-08-18?")
result3 = await assistant.run(task="Any event on 2025-08-20?")
print(result1.messages[-1].content)
print(result2.messages[-1].content)
print(result3.messages[-1].content)
```

```python
#SelectorGroupChat teams in Autogen
import asyncio
from autogen_agentchat.teams import SelectorGroupChat
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.conditions import MaxMessageTermination, TextMentionTermination
from autogen_agentchat.ui import Console

# ---------------------------
# 1. Planning Agent
# ---------------------------
planning_agent = AssistantAgent(
    name="Planning_Agent",
    description="Plans tasks into subtasks. First agent to engage.",
    model_client=model_client,
    system_message=(
        """You are the Planning Agent.
        Break down the complete task into smaller, manageable subtasks.
        Team members:
          - Web_Search_Agent: search for information
          - Data_Analyst_Agent: calculate percentage changes
        When assigning tasks, use format: <agent>:<task>.
        After completing all tasks, summarize findings and end with <TERMINATE>.
        """
    ),
)

# ---------------------------
# 2. Web Search Tool
# ---------------------------
def search_web_tool(query: str) -> str:
    if "2006-2007" in query:
        return """Here are the total points scored by Miami Heat players in the 2006-2007 season:
        Udonis Haslem: 844 points
        Dwyane Wade: 1397 points
        James Posey: 550 points"""
    
    elif "2007-2008" in query:
        return "In the 2007-2008 season, Dwyane Wade’s total rebounds were 214."
    
    elif "2008-2009" in query:
        return "In the 2008-2009 season, Dwyane Wade’s total rebounds were 398."
    
    else:
        return "No relevant data found."

web_search_agent = AssistantAgent(
    name="Web_Search_Agent",
    model_client=model_client,
    description="Searches for basketball stats and season details.",
    tools=[search_web_tool],
    system_message="You are the Web Search Agent. Use search_web_tool to fetch player and stats information."
)

# ---------------------------
# 3. Percentage Change Tool
# ---------------------------
def percentage_change_tool(start: float, end: float) -> float:
    if start == 0:
        return float("inf")  # avoid divide-by-zero
    return ((end - start) / start) * 100

# ---------------------------
# 4. Data Analyst Agent
# ---------------------------
data_analyst_agent = AssistantAgent(
    name="Data_Analyst_Agent", 
    model_client=model_client,
    description="Performs calculations and data analysis.",
    tools=[percentage_change_tool],
    system_message=(
        """You are the Data Analyst Agent.
        Use percentage_change_tool to analyze and calculate percentage changes.
        If you don’t have the data, ask for it."""
    ),
)

# ---------------------------
# 5. Termination Conditions
# ---------------------------
max_message_termination = MaxMessageTermination(20)
text_mention_termination = TextMentionTermination("TERMINATE")
combined_termination = text_mention_termination | max_message_termination

# ---------------------------
# 6. SelectorGroupChat
# ---------------------------
selector_prompt = """
Select the next agent to perform the task.

Current conversation history:
{history}

Available agents:
{participants}

Based on the conversation, select one agent to continue. 
The Planning_Agent should assign tasks first before others start working.
Only select one agent.
"""


Selector_team = SelectorGroupChat(
    participants=[planning_agent, web_search_agent, data_analyst_agent],
    selector_prompt=selector_prompt,
    model_client=model_client,
    termination_condition=combined_termination,
    allow_repeated_speaker=True,
)

# ---------------------------
# 7. Run
# ---------------------------
task = (
    "Who was the Miami Heat player with the highest points in the 2006-2007 season and what was the change in his total rebounds between 2007-2008 and 2008-2009?"
)
stream = Selector_team.run_stream(task=task)
await Console(stream)

```

```python

# Swarm: Travel Agent ↔ Flights Refunder (with user handoff)
# -----------------------------------------------------------
# Requirements:
#   pip install "autogen-agentchat>=0.4.0" "autogen-ext>=0.4.0"
#   export/set OPENAI_API_KEY first

import asyncio
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.teams import Swarm
from autogen_agentchat.conditions import HandoffTermination, TextMentionTermination
from autogen_agentchat.messages import HandoffMessage
from autogen_agentchat.ui import Console
from autogen_ext.models.openai import OpenAIChatCompletionClient

# ----- Model client (disable parallel tool calls for predictable handoffs) -----
model_client = OpenAIChatCompletionClient(
    model="gpt-4.1",              # or "gpt-4o" if you have access
    parallel_tool_calls=False
)

# -------------------- Tool: refund flight --------------------
def refund_flight(flight_id: str) -> str:
    """Refund a flight by reference/PNR and return a receipt text."""
    # (Replace this toy implementation with your real refund API call)
    return f"Flight {flight_id} refunded successfully. Refund receipt: R-{flight_id[-4:]}"

# -------------------- Agents with handoffs --------------------
travel_agent = AssistantAgent(
    name="travel_agent",
    model_client=model_client,
    # who this agent is allowed to hand off to
    handoffs=["flights_refunder", "user"],
    system_message=(
        "You are a helpful travel agent. "
        "The flights_refunder is responsible for actual refund execution. "
        "If you need information from the user (like PNR), first ask your question, "
        "then hand off to the user. "
        "When the overall task is complete, say TERMINATE."
    ),
)

flights_refunder = AssistantAgent(
    name="flights_refunder",
    model_client=model_client,
    handoffs=["travel_agent", "user"],
    tools=[refund_flight],
    system_message=(
        "You specialize in processing flight refunds. "
        "You only need the flight reference/PNR to refund a flight. "
        "Use the refund_flight tool when you have the PNR. "
        "If you need information, first ask, then hand off to the user. "
        "After refunding, hand off back to travel_agent to finalize."
    ),
)

# -------------------- Team (Swarm) + termination --------------------
termination = (
    HandoffTermination(target="user")   # pause when an agent hands off to the user
    | TextMentionTermination("TERMINATE")
)
team = Swarm(participants=[travel_agent, flights_refunder],
             termination_condition=termination)

# -------------------- Run helper: supports user-in-the-loop --------------------
async def run_refund_flow(initial_user_task: str):
    # Start with user's initial request
    task_result = await Console(team.run_stream(task=initial_user_task))
    last_msg = task_result.messages[-1]

    # If any agent handed off to "user", capture input and route back
    while isinstance(last_msg, HandoffMessage) and last_msg.target == "user":
        user_reply = input("User: ")
        # Route user's reply back to the requesting agent
        task_result = await Console(team.run_stream(
            task=HandoffMessage(
                source="user",                # the human
                target=last_msg.source,       # whoever asked the user
                content=user_reply
            )
        ))
        last_msg = task_result.messages[-1]

# -------------------- Kickoff --------------------
# Example: user wants a refund, agent should ask for PNR -> handoff to user
# Then refunder uses the tool -> hands back to travel_agent to close -> TERMINATE
await run_refund_flow("I need to refund my flight.")
await model_client.close()
```
