```python
class Burger:
    def __init__(self, bun, patty, sauce):
        self.bun = bun
        self.patty = patty
        self.sauce = sauce

    def describe(self):
     print(f"This burger has a {self.bun}bun, a {self.patty}patty, a {self.sauce}sauce.")
```

```python
my_burger = Burger ("sesame", "chicken", "mayo")
my_burger.describe()
```

```python
from pydantic import BaseModel

class Employee(BaseModel):
    id: int
    name: str
    salary: float

# Example usage
emp1 = Employee(id=101, name="John", salary=55000.50)
print(emp1)
```

```python
from pydantic import BaseModel, field_validator

class Student(BaseModel):
    name: str
    age: int
    grade: str

    # validator for age
    @field_validator("age")
    def check_age(cls, v):
        if v < 5:
            raise ValueError("Age must be at least 5")
        return v

# Example usage
student1 = Student(name="Alice", age=15, grade="10th")
print(student1)
```
